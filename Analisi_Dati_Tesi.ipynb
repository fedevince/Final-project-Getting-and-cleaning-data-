# Descriptive Analysis of the cohort


Table of demographics data:

```r
demographics <- cvid_data |> select(ID:DEATH_YEAR)
```

```r
demographics |>
  summarize(
    num_patients = n_distinct(ID),
    count_sex1 = sum(SEX == 1, na.rm = TRUE),
    prop_sex1 = sum(SEX == 1, na.rm = TRUE) / sum(!is.na(SEX)),
    count_sex0 = sum(SEX == 0, na.rm = TRUE),
    prop_sex0 = sum(SEX == 0, na.rm = TRUE) / sum(!is.na(SEX)),
    median_age_at_diagnosis = median(`AGE AT DIAGNOSIS_YEAR`, na.rm=TRUE),
    iqr_age_at_diagnosis = IQR(`AGE AT DIAGNOSIS_YEAR`, na.rm=TRUE),
    median_age = median(AGE_YEAR, na.rm = TRUE),
    iqr_age = IQR(AGE_YEAR, na.rm = TRUE),
    median_diagnostic_delay = median(`DIAGNOSTIC DELAY_YEAR`, na.rm=TRUE),
    iqr_diagnostic_delay = IQR(`DIAGNOSTIC DELAY_YEAR`, na.rm=TRUE)
    )
```

* Number of Patients: 149
* Sex 1: 81 patients (54.73%)
* Sex 0: 67 patients (45.27%)
* Median Age at Diagnosis: 39 years (IQR: 20.75)
* Median Age: 54 years (IQR: 23)
* Median Diagnostic Delay: 5 years (IQR: 12.75)



Percentage of Patients who died

```r
demographics |>
  summarize(
    total_patients = n_distinct(ID),
    patients_with_death = sum(DEATH == 1, na.rm = TRUE),
    percentage_death = (sum(DEATH == 1, na.rm = TRUE) / n_distinct(ID))
  )
```

## AMERATUNGA and IDDA scores

```r
cvid_data |>
  summarize(
    median_idda_score_at_follow_up = median(`IDDA SCORE \nAT FOLLOW UP`, na.rm = TRUE),
    iqr_idda_score_at_follow_up = IQR(`IDDA SCORE \nAT FOLLOW UP`, na.rm = TRUE),
    median_idda_score_at_diagnosis = median(`IDDA SCORE \nAT DIAGNOSIS`, na.rm = TRUE),
    iqr_idda_score_at_diagnosis = IQR(`IDDA SCORE \nAT DIAGNOSIS`, na.rm = TRUE),
    median_ameratunga_score_at_follow_up = median(`AMERATUNGA SCORE AT FOLLOW UP`, na.rm = TRUE),
    iqr_ameratunga_score_at_follow_up = IQR(`AMERATUNGA SCORE AT FOLLOW UP`, na.rm = TRUE),
    median_ameratunga_score_at_diagnosis = median(`AMERATUNGA SCORE AT DIAGNOSIS`, na.rm = TRUE),
    iqr_ameratunga_score_at_diagnosis = IQR(`AMERATUNGA SCORE AT DIAGNOSIS`, na.rm = TRUE)
  )
```
